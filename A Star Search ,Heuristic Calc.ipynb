```python
coords = {'S': (0, 0), 'A': (1, 2), 'B': (2, 1), 'G': (3, 3)}

def manhattan(node, goal):
    x1, y1 = coords[node]
    x2, y2 = coords[goal]
    return abs(x2 - x1) + abs(y2 - y1)

goal_node = 'G'
heuristic = {node: manhattan(node, goal_node) for node in coords}
```

A* SEARCH FULL (heuristic already given)

```python
import heapq

def Astar(graph,heuristic,start,goal):
    frontier = [(heuristic[start],start)]
    pathcost = {start : 0}
    came_from = {}
    visited = set()
    while frontier:
        h,current = heapq.heappop(frontier)
        if current == goal:
            return backtrack(came_from, current),pathcost[goal]
        if current in visited:
            continue
        visited.add(current)
        for neighbor,cost in graph[current].items():
            tent_cost = pathcost[current]+cost
            if tent_cost < pathcost.get(neighbor, float('inf')):
                pathcost[neighbor] = tent_cost
                came_from[neighbor] = current
                heapq.heappush(frontier, (tent_cost+heuristic[neighbor],neighbor))
    return None, float('inf')


def backtrack(came_from,current):
    path = [current]
    while current in came_from:
        current = came_from[current]
        path.append(current)
    return path[: : -1]



graph = {
        "S": {"A": 1, "B": 4},
        "A": {"S": 1, "B": 2, "C": 2},
        "B": {"S": 4, "A": 2, "G": 5},
        "C": {"A": 2, "G": 3},
        "G": {"B": 5, "C": 3},
    }
heuristic = {"S": 7, "A": 6, "B": 4, "C": 2, "G": 0}
 
path, cost = Astar(graph, heuristic, 'S', 'G')
print(path)
print(cost)
```
