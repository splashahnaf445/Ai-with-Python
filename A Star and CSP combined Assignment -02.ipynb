You are given a 2D grid representing a city map. Each cell is either open (.), a wall (#), or rough terrain (~, which costs more to cross). You must implement A* search to find the lowest-cost path from a start cell S to a goal cell G.

Write a function:

def astar(grid: list[str], start: tuple[int, int], goal: tuple[int, int]) -> tuple[list[tuple[int, int]], int]:
    """
    grid: list of strings, each string is one row of the map.
          '.' = open cell, cost 1 to enter
          '~' = rough terrain, cost 3 to enter
          '#' = wall, impassable
    start: (row, col) of the starting cell
    goal:  (row, col) of the goal cell

    Returns:
        (path, total_cost)
        path: list of (row, col) tuples from start to goal, inclusive.
              If no path exists, return ([], -1).
        total_cost: sum of entry costs along the path (start cell costs 0).
    """
Movement rules
You may move up, down, left, right (no diagonals).  // So, should use manhattan as eucledean finds diagonal distances!
You may not move into a # cell or off the grid.
Moving into . costs 1. Moving into ~ costs 3.
Heuristic requirement
Use the Manhattan distance between a cell and the goal as your heuristic function hNo. Explain in a comment above your heuristic function why Manhattan distance is admissible given the movement rules above.

Manhattan distance is admissible because movement is only up/down/left/right, so it gives the minimum number of movesneeded to reach the goal. Every move costs at least 1, so Manhattan distance can never overestimate the actual cost. (also eucledean finds diagonal distance, which is not the case here!)

```python
import heapq


def Astar(grid, start, goal):

    def heuristic(current):
        # Manhattan distance is admissible because movement is only
        # up/down/left/right, and every move costs at least 1.
        return abs(goal[0] - current[0]) + abs(goal[1] - current[1])

    frontier = [(heuristic(start), start)]

    pathcost = {
        start: 0
    }

    came_from = {}
    visited = set()

    rows = len(grid)
    cols = len(grid[0])

    directions = [
        (-1, 0),   # up
        (1, 0),    # down
        (0, -1),   # left
        (0, 1)     # right
    ]

    while frontier:

        f, current = heapq.heappop(frontier)

        if current == goal:
            return backtrack(came_from, current), pathcost[goal]

        if current in visited:
            continue

        visited.add(current)

        for dr, dc in directions:

            neighbor = (
                current[0] + dr,
                current[1] + dc
            )

            # Outside the grid
            if not (0 <= neighbor[0] < rows and
                    0 <= neighbor[1] < cols):
                continue

            # Wall
            if grid[neighbor[0]][neighbor[1]] == '#':
                continue

            # Cost of entering neighbor
            if grid[neighbor[0]][neighbor[1]] == '~':
                move_cost = 3
            else:
                move_cost = 1

            tent_cost = pathcost[current] + move_cost

            # Found a cheaper route to neighbor
            if tent_cost < pathcost.get(neighbor, float('inf')):

                pathcost[neighbor] = tent_cost
                came_from[neighbor] = current

                f_cost = tent_cost + heuristic(neighbor)

                heapq.heappush(frontier, (f_cost, neighbor))

    return [], -1


def backtrack(came_from, current):

    path = [current]

    while current in came_from:
        current = came_from[current]
        path.append(current)

    return path[::-1]


# Example

grid = [
    "S..#....",
    ".##.~~..",
    "....~.#.",
    "..#..#..",
    "....#..G",
]

start = (0, 0)
goal = (4, 7)

path, cost = Astar(grid, start, goal)

print("Path:", path)
print("Cost:", cost)
```
